```python
import torch
import torch.nn as nn
```

```python
xs = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

ys = torch.tensor([[-1.],[1.],[1.],[-1.]]) # -> Ye expected output hai.
```

```python
# Ye kaam karega same MLP ka:-
model = nn.Sequential(
    nn.Linear(2, 4),  # -> Takes a 2D input vector(x1 , x2) and maps it to a 4D hidden vector 
    nn.Tanh(), 
    nn.Linear(4, 1), # -> Output Layer (4 hidden nodes se 1 output banata hai)
    nn.Tanh() # -> final value par tanh lagata hai
)
```

```python
# torch.optim.SGD -> loss ko kam↓ karna iska kaam hai.
# model.parameters -> model ke saare weights aur biases. lr -> learning rate.
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
```

```python
optimizer.zero_grad()

ypred = model(xs) # -> Forward Pass (Prediction)
loss = ((ypred - ys) ** 2).sum() # -> loss calculate kar raha hai

loss.backward() # -> backward propagation
optimizer.step() # -> Weights Update Karna
```

```python
# Model ko train karne ke liye loop (2000 epochs)
epochs = 2000

for epoch in range(epochs):
    optimizer.zero_grad()  # 1. Gradients reset
    
    ypred = model(xs)      # 2. Forward pass
    loss = ((ypred - ys) ** 2).sum()  # 3. Loss calculation
    
    loss.backward()        # 4. Backward pass
    optimizer.step()       # 5. Weights update
    
    # Har 200 steps baad loss print karo taaki progress dikhe
    if (epoch + 1) % 200 == 0:
        print(f"Epoch {epoch + 1}/{epochs} | Loss: {loss.item():.4f}")
```
